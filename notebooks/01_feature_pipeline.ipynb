```python
import joblib
from loan_classifier.features import (
    load_raw_data, engineer_features,
    split_data, build_preprocessor, get_feature_groups
)
```

# 1. Load and engineer

```python
df = load_raw_data("../data/raw/trainData.csv")
df = engineer_features(df)
```

# 2. Split FIRST before any fitting

```python
X_train, X_val, X_test, y_train, y_val, y_test = split_data(df)
```

# 3. Build preprocessor and fit on X_train ONLY

```python
preprocessor = build_preprocessor()
preprocessor.fit(X_train)
```

# 4. Save — this same object is loaded in the Inference Pipeline

```python
joblib.dump(preprocessor, "../models/preprocessor.joblib")
```

# 5. Save splits for the Training Pipeline

```python
X_train.to_csv("../data/processed/X_train.csv", index=False)
X_val.to_csv("../data/processed/X_val.csv", index=False)
X_test.to_csv("../data/processed/X_test.csv", index=False)
y_train.to_csv("../data/processed/y_train.csv", index=False)
y_val.to_csv("../data/processed/y_val.csv", index=False)
y_test.to_csv("../data/processed/y_test.csv", index=False)
```
